## Interpretation boundaries

- P01 is a synthetic CCD-like reconstruction constrained by published FWHM and temperature tables; it is not raw-frame recovery.
- P02 physical mode identification is performed by explicit spatial Fourier projection on the synthetic 12-channel signals; SVD component number is not treated as physical mode number.
- P02 HXR spectra are count/mean/total-energy constrained synthetic spectra; detector response is not reconstructed.
- P03 uncertainty is a sensitivity/bootstrapping result on synthetic current ratios.
- P04/P05 intermediate fluctuation and field traces are model-generated; reported transport/effect values remain publication constraints, not discovered outputs from raw data.
- P06 frequency comparison is performed on reconstructed multichannel Mirnov data and preserves the paper's stated frequency and timing inconsistency as a limitation.
