## Interpretation boundaries

- **P01:** scientific analysis must use the author-supplied laboratory measurements. The legacy synthetic CCD-like spectrum is retained only as a software-test scaffold; its comparison with publication values is withdrawn and must not be displayed or reported as an experimental discrepancy. Dataset A and Dataset B remain separate.
- **P02:** physical mode identification uses explicit spatial Fourier projection on synthetic 12-channel signals; SVD component number is not treated as physical mode number. HXR spectra are synthetic and constrained by reported counts/energies; detector response is not reconstructed.
- **P03:** uncertainty is model sensitivity on synthetic current ratios; the original Mach time series is unavailable.
- **P04:** text-level targets remain separate from approximate Dataset B values; intermediate traces are synthetic.
- **P05:** numerical effects are parameterization targets, not independent validation.
- **P06:** synthetic signals use text-level approximately 44 kHz activity; pressure-specific frequencies and digitized peaks remain approximate. The source timing inconsistency is preserved.

Passing software tests establishes software execution/reproducibility only, not independent experimental validation.
