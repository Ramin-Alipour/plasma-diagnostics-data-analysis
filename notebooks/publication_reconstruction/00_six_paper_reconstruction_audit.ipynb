## Interpretation boundaries- P01 is a synthetic CCD-like reconstruction constrained by published FWHM data and the paper's Eq. (3)/(11) resolution model; the resulting temperature comparison is a computational consistency check, not raw-frame recovery or independent validation.- P02 physical mode identification is performed by explicit spatial Fourier projection on the synthetic 12-channel signals; SVD component number is not treated as physical mode number.- P02 HXR spectra are count/mean/total-energy constrained synthetic spectra; detector response is not reconstructed.- P03 uncertainty is a sensitivity/bootstrapping result on synthetic current ratios.- P04 text-level targets are kept separate from approximate Dataset B absolute values; all intermediate traces are synthetic. P05 percentage effects are parameterization targets, not independent validation. P06 synthetic signals use the text-level approximately 44 kHz target; pressure-specific frequencies and digitized peaks remain approximate.- P06 frequency comparison is performed on reconstructed multichannel Mirnov data and preserves the paper's stated frequency and timing inconsistency as a limitation.
